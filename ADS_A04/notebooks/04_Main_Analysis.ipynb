### Data Cleaning & Feature Engineering Pipeline 🧹
> Running custom text cleaners, joining tables, extracting features, and saving the final state into our processed data folder. 💾

```python
PROCESSED_DATA_DIR = '../data/processed/'
PROCESSED_DATA_PATH = os.path.join(PROCESSED_DATA_DIR, 'tracks.csv')

def validate_missing_thresholds(df: pd.DataFrame) -> pd.DataFrame:
    """Drops columns that contain entirely missing fields or duplicates."""
    df = df.drop_duplicates()
    return df

def clean_custom_placeholders(df: pd.DataFrame) -> pd.DataFrame:
    """Replaces text variations of empty variables with proper null objects safely."""
    return df.replace([r'^null$', r'^\[null\]$', '^unknown$', r'^\?$'], np.nan, regex=True)

# 1. Clean and join via deployed modules
t_flat, a_flat, u_flat = flatten_and_clean(df_reg_tracks, df_reg_artists, df_reg_users)
t_id, a_id, u_id = simplify_all_ids(t_flat, a_flat, u_flat)
processed_base_df = aggregate_track_features(t_id, a_id, u_id)

# 2. Functional cleaning chain
cleaned_dataset = (
    processed_base_df
    .pipe(validate_missing_thresholds)
    .pipe(clean_custom_placeholders)
)

# 3. Create target binarization threshold FIRST! 🎯
popular_reaction_threshold = 15
cleaned_dataset['is_popular'] = (cleaned_dataset['total_reactions'] > popular_reaction_threshold).astype(int)

# 4. Now filter the columns using cleaned_dataset ✂️
keep_columns = [
    'track_id', 'num_senders', 'total_reactions', 'artists_total_likes', 
    'artists_total_dislikes', 'artists_total_reactions', 'senders_total_likes', 
    'senders_total_dislikes', 'senders_total_reactions', 'senders_total_likes_received', 
    'senders_total_dislikes_received', 'senders_total_reactions_received', 
    'senders_total_liked_users', 'senders_total_disliked_users', 'senders_total_reacted_users', 
    'senders_total_users_liked', 'senders_total_users_disliked', 'senders_total_users_reacted', 
    'senders_total_uploaded_tracks', 'is_popular' 
]

available_columns = [col for col in keep_columns if col in cleaned_dataset.columns]
cleaned_dataset = cleaned_dataset[available_columns]

# 5. Export to processed folder
os.makedirs(PROCESSED_DATA_DIR, exist_ok=True)
cleaned_dataset.to_csv(PROCESSED_DATA_PATH, index=False)
print(f"💾 Cleaned and engineered dataset saved to: {PROCESSED_DATA_PATH}\n")

# 6. Reload the freshly processed dataset
df_processed = pd.read_csv(PROCESSED_DATA_PATH)
print(f"🔄 Processed data reloaded successfully! Shape: {df_processed.shape}")
display(df_processed.head())
```

### Preprocessing Pipeline & Missing Data Handling 🧬
> Setting up custom transformers to test Mean, Median, KNN, and Iterative imputation directly inside our processing pipeline. 🧪

```python
# Extract features and targets from reloaded data
X = df_processed.drop(columns=['is_popular', 'track_id', 'total_reactions'], errors='ignore')
y = df_processed['is_popular']

numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()

def generate_preprocessor(imputation_strategy='median'):
    """Generates structural transformers matching selected missing data engines."""
    if imputation_strategy == 'mean':
        num_imputer = SimpleImputer(strategy='mean')
    elif imputation_strategy == 'median':
        num_imputer = SimpleImputer(strategy='median')
    elif imputation_strategy == 'knn':
        num_imputer = KNNImputer(n_neighbors=5)
    elif imputation_strategy == 'iterative':
        num_imputer = IterativeImputer(max_iter=10, random_state=42)
        
    numeric_transformer = Pipeline(steps=[
        ('imputer', num_imputer),
        ('scaler', StandardScaler())
    ])
    
    categorical_transformer = Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
    ])
    
    return ColumnTransformer(transformers=[
        ('num', numeric_transformer, numerical_cols),
        ('cat', categorical_transformer, categorical_cols)
    ])

# Matrix structural verification checks
for strategy in ['mean', 'median', 'knn', 'iterative']:
    transformer_unit = generate_preprocessor(strategy)
    mock_transform = transformer_unit.fit_transform(X)
    print(f"✨ Imputation strategy '{strategy}' processed matrix layout shape: {mock_transform.shape}")
```

### End-to-End Trainable Model 🤖
> Attaching the final random forest classifier to our preprocessing flow and saving the entire trainable engine. 🎯

```python
PROCESSED_MODEL_DIR = '../data/models/'
PROCESSED_MODEL_PATH = os.path.join(PROCESSED_MODEL_DIR, 'tracks.pkl')

# 1. Split data matrices
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# 2. Task 3 Validation: Compare all 4 imputation techniques systematically
imputation_strategies = ['mean', 'median', 'knn', 'iterative']
best_strategy = 'median'
best_f1 = 0

print("🕵️‍♂️ Evaluating Task 3 Imputation Strategies...")
print("-" * 50)
for strategy in imputation_strategies:
    test_pipeline = Pipeline(steps=[
        ('preprocessor', generate_preprocessor(imputation_strategy=strategy)),
        ('classifier', RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42))
    ])
    test_pipeline.fit(X_train, y_train)
    val_preds = test_pipeline.predict(X_test)
    rep = classification_report(y_test, val_preds, output_dict=True)
    class_1_f1 = rep['1']['f1-score']
    print(f"🔹 Strategy: {strategy:<10} | Overall Accuracy: {rep['accuracy']:.4f} | Class 1 F1-Score: {class_1_f1:.4f}")
    
    if class_1_f1 > best_f1:
        best_f1 = class_1_f1
        best_strategy = strategy

print("-" * 50)
print(f"🏆 Best Imputation Strategy Selected: {best_strategy.upper()}\n")

# 3. Re-train the final Unified Pipeline using the best imputation strategy
PROCESSED_MODEL_DIR = '../data/models/'
PROCESSED_MODEL_PATH = os.path.join(PROCESSED_MODEL_DIR, 'tracks.pkl')

end_to_end_pipeline = Pipeline(steps=[
    ('preprocessor', generate_preprocessor(imputation_strategy=best_strategy)),
    ('classifier', RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42))
])

print("🏋️ Training the final machine learning pipeline...")
end_to_end_pipeline.fit(X_train, y_train)

# 4. Evaluate and generate visualizations
predictions = end_to_end_pipeline.predict(X_test)
probabilities = end_to_end_pipeline.predict_proba(X_test)[:, 1]

print("\n📋 Out-of-Sample Performance Report:")
print(classification_report(y_test, predictions))

# 5. Serialize model to disk
os.makedirs(os.path.dirname(PROCESSED_MODEL_PATH), exist_ok=True)
joblib.dump(end_to_end_pipeline, PROCESSED_MODEL_PATH)
print(f"📦 Pipeline successfully serialized to disk: {PROCESSED_MODEL_PATH}")

# 6. Plot performance metrics side-by-side
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# Confusion Matrix Heatmap
cm = confusion_matrix(y_test, predictions)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax[0], 
            xticklabels=['Not Popular (0)', 'Popular (1)'], 
            yticklabels=['Not Popular (0)', 'Popular (1)'],
            linecolor='black', linewidths=0.5)
ax[0].set_title('Confusion Matrix Heatmap', fontsize=12, fontweight='bold')
ax[0].set_xlabel('Predicted Label')
ax[0].set_ylabel('True Label')

# ROC Curve
fpr, tpr, _ = roc_curve(y_test, probabilities)
roc_auc = auc(fpr, tpr)
ax[1].plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.4f})')
ax[1].plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
ax[1].set_xlim([0.0, 1.0])
ax[1].set_ylim([0.0, 1.05])
ax[1].set_xlabel('False Positive Rate')
ax[1].set_ylabel('True Positive Rate')
ax[1].set_title('Receiver Operating Characteristic (ROC)', fontsize=12, fontweight='bold')
ax[1].legend(loc="lower right")
ax[1].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
