```python
import pandas as pd
import re
from pathlib import Path

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------
BASE_DIR = Path("./data")   # change to your actual folder
OUTPUT_DIR = BASE_DIR               # overwrite, or set to a different folder

# Which columns in each table need fixing (list of column names)
FIX_COLUMNS = {
    "artists": ["reactions"],
    "tracks":  ["reactions"],       # change if column name is different
    "users":   ["reacted_tracks", "reactions_received", "reacted_users", "users_reacted"]
}

# Mapping for broken values -> correct integer string
BROKEN_VALUE_MAP = {
    "4\u200d5": "5",      # heart + ZWJ + fire (without Variation Selector)
    "4â€5": "5",         # same, but mojibake representation (just in case)
    "☃": "2",            # snowman (U+2603) – was missing from mapping
}

# ------------------------------------------------------------
# 2. HELPER FUNCTIONS
# ------------------------------------------------------------
def fix_value(val):
    """Return a cleaned value; if val is numeric string, return it unchanged."""
    if not isinstance(val, str):
        return val
    val = val.strip()
    # If it's already a valid integer (possibly with '-'), return it
    if re.match(r'^-?\d+$', val):
        return val
    # Try to apply known replacements
    for broken, fixed in BROKEN_VALUE_MAP.items():
        if broken in val:
            val = val.replace(broken, fixed)
            # If after replacement it becomes a pure integer, return immediately
            if re.match(r'^-?\d+$', val):
                return val
    # If still not numeric, keep as-is (but we'll log a warning)
    return val

def fix_id_value_string(text):
    """
    Given a string like "486:5, 655:5, 725:2, ...",
    split into pairs, clean each value, and rebuild.
    Returns the fixed string and a dict of changes (broken_pattern -> count).
    """
    if not isinstance(text, str):
        return text, {}
    pairs = [p.strip() for p in text.split(",") if p.strip()]
    fixed_pairs = []
    changes = {}
    for pair in pairs:
        if ":" not in pair:
            # malformed – skip or keep as is
            fixed_pairs.append(pair)
            continue
        parts = pair.split(":", 1)
        if len(parts) != 2:
            fixed_pairs.append(pair)
            continue
        key, value = parts[0].strip(), parts[1].strip()
        original_value = value
        new_value = fix_value(value)
        if new_value != original_value:
            # count the change
            changes[original_value] = changes.get(original_value, 0) + 1
        fixed_pairs.append(f"{key}:{new_value}")
    return ", ".join(fixed_pairs), changes

def fix_column(df, col_name):
    """
    Apply fix_id_value_string to every cell in the given column.
    Returns a new Series, and a dict summarizing total changes.
    """
    if col_name not in df.columns:
        print(f"  ⚠️ Column '{col_name}' not found – skipping.")
        return df[col_name], {}
    total_changes = {}
    new_series = []
    for cell in df[col_name]:
        if pd.isna(cell):
            new_series.append(cell)
            continue
        fixed_str, changes = fix_id_value_string(str(cell))
        new_series.append(fixed_str)
        for k, v in changes.items():
            total_changes[k] = total_changes.get(k, 0) + v
    return pd.Series(new_series, index=df.index), total_changes

# ------------------------------------------------------------
# 3. PROCESS EACH TABLE
# ------------------------------------------------------------
def process_table(table_name, df):
    print(f"\n📁 Processing table: {table_name}")
    cols = FIX_COLUMNS.get(table_name, [])
    if not cols:
        print("  No columns to fix.")
        return df, {}
    grand_total = {}
    for col in cols:
        print(f"  🔧 Fixing column: '{col}'")
        new_col, changes = fix_column(df, col)
        df[col] = new_col
        if changes:
            for broken, count in changes.items():
                print(f"    - replaced '{broken}' → '{BROKEN_VALUE_MAP.get(broken, '?')}' in {count} entries")
                grand_total[broken] = grand_total.get(broken, 0) + count
        else:
            print("    No changes needed.")
    return df, grand_total

# ------------------------------------------------------------
# 4. LOAD, FIX, SAVE
# ------------------------------------------------------------
# Load the three tables (adjust filenames as needed)
tables = {}
for name in ["artists", "tracks", "users"]:
    file_path = BASE_DIR / f"{name}.csv"   # adjust to your actual filenames
    if not file_path.exists():
        print(f"❌ File not found: {file_path}")
        continue
    try:
        df = pd.read_csv(file_path, encoding='utf-8')
        tables[name] = df
        print(f"✅ Loaded {name} ({len(df)} rows)")
    except Exception as e:
        print(f"❌ Failed to load {name}: {e}")

# Process each loaded table
for name, df in tables.items():
    fixed_df, changes = process_table(name, df)
    tables[name] = fixed_df
    # Drop 'language' column if it exists in users
    if name == "users" and "language" in fixed_df.columns:
        fixed_df = fixed_df.drop(columns=["language"])
        print("  🗑️ Dropped 'language' column.")
        tables[name] = fixed_df

# Save the fixed tables
for name, df in tables.items():
    out_path = OUTPUT_DIR / f"{name}_fixed.csv"
    df.to_csv(out_path, index=False, encoding='utf-8')
    print(f"💾 Saved fixed {name} to {out_path}")

print("\n✅ All done.")
```

```python
import pandas as pd
import re

# ----- CONFIG -----
# Paths to your current "fixed" CSV files (the ones that still contain anomalies)
INPUT_FILES = {
    "artists": "data/artists_fixed.csv",
    "tracks":  "data/tracks_fixed.csv",
    "users":   "data/users_fixed.csv"
}

# Columns that contain "id:value" pairs
COLUMNS = {
    "artists": ["reactions"],
    "tracks":  ["reactions"],
    "users":   ["reacted_tracks", "reactions_received", "reacted_users", "users_reacted"]
}

# ----- SCORES FOR MISSING EMOJIS / STRINGS (UPDATED) -----
MISSING_SCORES = {
    # High Energy / Intense Positive (+5)
    "🥰": '5',
    "😍": '5',
    "🤣": '5',
    "😱": '5',
    "CustomEmoji": '5',
    
    # Positive / Enjoyment (+3 to +4)
    "🤗": '4',
    "🙏": '3',
    "🫡": '3',
    "🐳": '3',           # ← promoted to 3
    "✍": '4',            # ← promoted to 4
    
    # Mild Positive (+1 to +2)
    "🕊": '2',
    "🎅": '2',
    "🙈": '2',
    "👀": '1',
    "👻": '1',
    "👾": '1',
    "🗿": '1',
    "💊": '1',
    
    # Neutral / Slightly Negative (-1)
    "🤷\u200d♂": '-1',
    "😴": '-1',
    
    # Strong Negative (-5)
    "💩": '-5',          # ← now very negative
}

# ----- FUNCTION TO REPLACE IN A SINGLE COLUMN -----
def replace_missing_emojis_in_column(df, col):
    """
    For each cell in the column, replace any occurrence of the missing emoji keys
    with their integer scores.
    Returns a tuple (new_series, replacement_counts) where replacement_counts
    is a dict mapping emoji -> number of replacements.
    """
    if col not in df.columns:
        return df[col], {}
    
    # Sort keys by length descending to handle ZWJ sequences correctly
    sorted_keys = sorted(MISSING_SCORES.keys(), key=len, reverse=True)
    
    replacement_counts = {k: 0 for k in MISSING_SCORES}
    new_values = []
    
    for cell in df[col]:
        if pd.isna(cell) or not isinstance(cell, str):
            new_values.append(cell)
            continue
        
        text = cell
        for emoji in sorted_keys:
            score = MISSING_SCORES[emoji]
            count = text.count(emoji)
            if count > 0:
                replacement_counts[emoji] += count
                text = text.replace(emoji, score)
        new_values.append(text)
    
    counts = {k: v for k, v in replacement_counts.items() if v > 0}
    return pd.Series(new_values, index=df.index), counts

# ----- PROCESS EACH TABLE -----
total_replacements = {}

for table_name, file_path in INPUT_FILES.items():
    print(f"\n📁 Processing {table_name} from {file_path}")
    df = pd.read_csv(file_path, encoding='utf-8')
    
    cols = COLUMNS.get(table_name, [])
    table_counts = {}
    
    for col in cols:
        if col not in df.columns:
            print(f"  ⚠️ Column '{col}' not found – skipping")
            continue
        print(f"  🔧 Fixing column: '{col}'")
        new_col, counts = replace_missing_emojis_in_column(df, col)
        df[col] = new_col
        if counts:
            for emoji, cnt in counts.items():
                print(f"    - replaced {repr(emoji)} → {MISSING_SCORES[emoji]} in {cnt} pairs")
                table_counts[emoji] = table_counts.get(emoji, 0) + cnt
        else:
            print("    No replacements needed.")
    
    # Save the cleaned table (you can change suffix)
    out_path = file_path.replace("_fixed.csv", "_clean.csv")
    df.to_csv(out_path, index=False, encoding='utf-8')
    print(f"💾 Saved cleaned {table_name} to {out_path}")
    
    for emoji, cnt in table_counts.items():
        total_replacements[emoji] = total_replacements.get(emoji, 0) + cnt

# ----- SUMMARY -----
print("\n" + "="*60)
print("📊 SUMMARY OF ALL REPLACEMENTS ACROSS ALL TABLES")
for emoji, cnt in sorted(total_replacements.items(), key=lambda x: -x[1]):
    print(f"  {cnt:5} ×  {repr(emoji)} → {MISSING_SCORES[emoji]}")
print("="*60)
print("✅ All done! Now you have fully integer-encoded tables.")
```

```python
import pandas as pd

# ----- LOAD CLEAN FILES -----
tracks = pd.read_csv("data/tracks_clean.csv")
artists = pd.read_csv("data/artists_clean.csv")

# ----- EXPLODE ARTIST NAMES CORRECTLY -----
# Keep only the columns we need
tracks_exploded = tracks[['artist_names', 'reactions']].copy()
# Split the comma-separated names into a list
tracks_exploded['artist_names'] = tracks_exploded['artist_names'].str.split(',')
# Explode the lists into separate rows
tracks_exploded = tracks_exploded.explode('artist_names')
# Strip whitespace and rename
tracks_exploded['artist_name'] = tracks_exploded['artist_names'].str.strip()
# Drop the temporary column and reset index
tracks_exploded = tracks_exploded.drop(columns=['artist_names']).reset_index(drop=True)
# Remove empty names
tracks_exploded = tracks_exploded[tracks_exploded['artist_name'].notna() & (tracks_exploded['artist_name'] != '')]

# ----- HELPER: EXTRACT SCORES -----
def extract_scores(reactions_str):
    if pd.isna(reactions_str) or not isinstance(reactions_str, str):
        return set()
    scores = set()
    for pair in reactions_str.split(','):
        pair = pair.strip()
        if ':' in pair:
            _, score = pair.split(':', 1)
            score = score.strip()
            if score.lstrip('-').isdigit():
                scores.add(int(score))
    return scores

# ----- 1. ARTIST EXISTENCE CHECK -----
track_artists = set(tracks_exploded['artist_name'])
artist_table_artists = set(artists['artist_name'].dropna())

print("📊 ARTIST EXISTENCE CHECK")
print(f"  Unique artists in tracks: {len(track_artists)}")
print(f"  Unique artists in artists table: {len(artist_table_artists)}")
missing = track_artists - artist_table_artists
if missing:
    print(f"  ❌ Artists in tracks but NOT in artists table: {missing}")
else:
    print("  ✅ All artists in tracks are present in the artists table.")

# ----- 2. COLLECT SCORE SETS PER ARTIST -----
# Track scores
artist_track_scores = {}
for artist in track_artists:
    rows = tracks_exploded[tracks_exploded['artist_name'] == artist]
    all_scores = set()
    for _, row in rows.iterrows():
        all_scores.update(extract_scores(row['reactions']))
    artist_track_scores[artist] = all_scores

# Profile scores (from artists table)
artist_profile_scores = {}
for _, row in artists.iterrows():
    artist = row['artist_name']
    if artist in track_artists:
        artist_profile_scores[artist] = extract_scores(row['reactions'])

# ----- 3. COMPARE SCORE SETS -----
print("\n🔍 REACTION SCORE CONSISTENCY (only artists with reactions shown)\n")

discrepancies = []
for artist in sorted(track_artists):
    if artist not in artist_profile_scores:
        print(f"⚠️ {artist}: missing from artists table (skipped)")
        continue

    track_scores = artist_track_scores.get(artist, set())
    profile_scores = artist_profile_scores.get(artist, set())

    if not track_scores and not profile_scores:
        continue

    if track_scores == profile_scores:
        print(f"✅ {artist}: scores match → {track_scores}")
    else:
        print(f"❌ {artist}:")
        print(f"     Track scores: {track_scores}")
        print(f"     Profile scores: {profile_scores}")
        if track_scores.issubset(profile_scores):
            print("     → Track scores are a SUBSET of profile scores.")
        elif profile_scores.issubset(track_scores):
            print("     → Profile scores are a SUBSET of track scores.")
        else:
            print("     → Neither is a subset – they are DIFFERENT data sources.")
        discrepancies.append(artist)

# ----- 4. GLOBAL SUMMARY -----
all_track_scores = set().union(*artist_track_scores.values()) if artist_track_scores else set()
all_profile_scores = set().union(*artist_profile_scores.values()) if artist_profile_scores else set()

print("\n📈 GLOBAL SCORE SETS")
print(f"  All scores in track reactions: {all_track_scores}")
print(f"  All scores in artist profiles: {all_profile_scores}")
print(f"  Scores unique to tracks: {all_track_scores - all_profile_scores}")
print(f"  Scores unique to profiles: {all_profile_scores - all_track_scores}")

if discrepancies:
    print(f"\n⚠️ Found {len(discrepancies)} artists with mismatched score sets.")
    print("   This confirms that artist profile reactions are NOT derived from track reactions.")
else:
    print("\n✅ All artists have consistent score sets (no discrepancies).")
```

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from sklearn.preprocessing import LabelEncoder
from scipy.sparse import csr_matrix, coo_matrix
import random
from collections import defaultdict

# ---------------------------------------------------------------------
# 0. Configuration
# ---------------------------------------------------------------------
FIXED_LIKE_ENERGY = 1           # thumb up
FIXED_DISLIKE_ENERGY = -2       # thumb down
ALPHA = 0.5                     # debiasing strength (reduced)
EMBEDDING_DIM = 50              # latent dimensions
BATCH_SIZE = 1024
EPOCHS = 10
LR = 0.001
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
TEST_PERCENT = 0.2
NUM_NEGATIVE = 8                # increased negatives per positive

# ---------------------------------------------------------------------
# 1. Parse reactions from tracks
# ---------------------------------------------------------------------
def parse_track_reactions(df_tracks):
    interactions = []
    for _, row in df_tracks.iterrows():
        track_id = row['track_id']
        sender_ids = str(row['sender_chat_ids'])
        sender_list = [int(x.strip()) for x in sender_ids.split(',') if x.strip()]
        sender = sender_list[0] if len(sender_list) == 1 else -1

        if pd.notna(row['likes']):
            for uid in str(row['likes']).split(','):
                uid = int(uid.strip())
                interactions.append((uid, track_id, FIXED_LIKE_ENERGY, sender))

        if pd.notna(row['dislikes']):
            for uid in str(row['dislikes']).split(','):
                uid = int(uid.strip())
                interactions.append((uid, track_id, FIXED_DISLIKE_ENERGY, sender))

        if pd.notna(row['reactions']):
            for token in str(row['reactions']).split(','):
                token = token.strip()
                if ':' not in token:
                    continue
                uid_str, energy_str = token.split(':')
                uid = int(uid_str)
                energy = int(energy_str)
                interactions.append((uid, track_id, energy, sender))

    df = pd.DataFrame(interactions, columns=['user_id', 'track_id', 'energy', 'sender_id'])
    return df

# ---------------------------------------------------------------------
# 2. Compute poster affinity
# ---------------------------------------------------------------------
def compute_poster_affinity(df_int):
    pos = df_int[df_int['energy'] > 0]
    if pos.empty:
        return pd.DataFrame(columns=['user_id', 'sender_id', 'affinity'])
    aff = pos.groupby(['user_id', 'sender_id'])['energy'].mean().reset_index()
    aff.rename(columns={'energy': 'affinity'}, inplace=True)
    return aff

# ---------------------------------------------------------------------
# 3. Build debiased matrix + artist features
# ---------------------------------------------------------------------
def build_debiased_matrix(df_int, df_aff, df_tracks, alpha=ALPHA):
    df = df_int.merge(df_aff, on=['user_id', 'sender_id'], how='left')
    df['affinity'] = df['affinity'].fillna(0)
    df['weight'] = (df['energy'] - alpha * df['affinity']).clip(lower=0)
    df_pos = df[df['weight'] > 0].copy()

    user_enc = LabelEncoder()
    item_enc = LabelEncoder()
    df_pos['user_idx'] = user_enc.fit_transform(df_pos['user_id'])
    df_pos['item_idx'] = item_enc.fit_transform(df_pos['track_id'])

    n_users = len(user_enc.classes_)
    n_items = len(item_enc.classes_)

    mat = csr_matrix((df_pos['weight'], (df_pos['user_idx'], df_pos['item_idx'])),
                     shape=(n_users, n_items))

    # ---- Build artist features (one‑hot) ----
    # Map each track to its artist string (only for tracks that survived encoding)
    track_to_artist = df_tracks.set_index('track_id')['artist_names'].to_dict()
    artist_enc = LabelEncoder()
    all_artists = df_tracks['artist_names'].unique()
    artist_enc.fit(all_artists)

    data, row_idx, col_idx = [], [], []
    for track_id, artist in track_to_artist.items():
        if track_id in item_enc.classes_:
            i = item_enc.transform([track_id])[0]
            a = artist_enc.transform([artist])[0]
            data.append(1)
            row_idx.append(i)
            col_idx.append(a)

    item_features = csr_matrix((data, (row_idx, col_idx)),
                               shape=(n_items, len(artist_enc.classes_)))

    return mat, user_enc, item_enc, df_pos, item_features

# ---------------------------------------------------------------------
# 4. BPR Dataset (with optional popularity‑weighted sampling later)
# ---------------------------------------------------------------------
class BPRDataset(Dataset):
    def __init__(self, interactions_df, n_items, num_negative=8):
        self.users = torch.LongTensor(interactions_df['user_idx'].values)
        self.pos_items = torch.LongTensor(interactions_df['item_idx'].values)
        self.n_items = n_items
        self.num_negative = num_negative

    def __len__(self):
        return len(self.users)

    def __getitem__(self, idx):
        user = self.users[idx]
        pos = self.pos_items[idx]
        negs = []
        while len(negs) < self.num_negative:
            neg = random.randint(0, self.n_items - 1)
            if neg != pos:
                negs.append(neg)
        return user, pos, torch.LongTensor(negs)

# ---------------------------------------------------------------------
# 5. Improved model with user bias and item features
# ---------------------------------------------------------------------
class HybridMF(nn.Module):
    def __init__(self, n_users, n_items, emb_dim, item_feature_dim=None):
        super().__init__()
        self.user_emb = nn.Embedding(n_users, emb_dim)
        self.item_emb = nn.Embedding(n_items, emb_dim)
        self.user_bias = nn.Embedding(n_users, 1)
        self.item_bias = nn.Embedding(n_items, 1)

        nn.init.xavier_uniform_(self.user_emb.weight)
        nn.init.xavier_uniform_(self.item_emb.weight)
        nn.init.zeros_(self.user_bias.weight)
        nn.init.zeros_(self.item_bias.weight)

        self.use_features = item_feature_dim is not None
        if self.use_features:
            self.feature_proj = nn.Linear(item_feature_dim, emb_dim, bias=False)

    def forward(self, user_idx, item_idx, item_features=None):
        u = self.user_emb(user_idx)
        i = self.item_emb(item_idx)
        if self.use_features and item_features is not None:
            # item_features: (batch, item_feature_dim)
            i = i + self.feature_proj(item_features)
        dot = (u * i).sum(dim=-1)
        u_bias = self.user_bias(user_idx).squeeze(-1)
        i_bias = self.item_bias(item_idx).squeeze(-1)
        return dot + u_bias + i_bias

# ---------------------------------------------------------------------
# 6. BPR loss
# ---------------------------------------------------------------------
def bpr_loss(model, user, pos, negs, pos_features=None, neg_features=None):
    pos_score = model(user, pos, pos_features)
    # negs: (batch, num_neg)
    batch_size, num_neg = negs.shape
    user_exp = user.unsqueeze(1).expand(-1, num_neg).reshape(-1)
    neg_flat = negs.reshape(-1)
    neg_f_flat = neg_features.reshape(-1, neg_features.size(-1)) if neg_features is not None else None
    neg_scores = model(user_exp, neg_flat, neg_f_flat).view(batch_size, num_neg)
    diff = pos_score.unsqueeze(1) - neg_scores
    return -torch.log(torch.sigmoid(diff) + 1e-10).mean()

# ---------------------------------------------------------------------
# 7. Training
# ---------------------------------------------------------------------
def train_epoch(model, dataloader, optimizer, item_features_tensor, device):
    model.train()
    total_loss = 0
    for user, pos, negs in dataloader:
        user, pos, negs = user.to(device), pos.to(device), negs.to(device)
        # Get feature vectors for pos and neg items
        pos_feat = item_features_tensor[pos] if item_features_tensor is not None else None
        neg_feat = item_features_tensor[negs] if item_features_tensor is not None else None
        optimizer.zero_grad()
        loss = bpr_loss(model, user, pos, negs, pos_feat, neg_feat)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    return total_loss / len(dataloader)

# ---------------------------------------------------------------------
# 8. Per‑user evaluation
# ---------------------------------------------------------------------
def evaluate_per_user(model, test_mat, train_mat, k=10):
    """
    Compute recall@k and precision@k for each user, then average over users.
    """
    model.eval()
    with torch.no_grad():
        item_emb = model.item_emb.weight
        user_emb = model.user_emb.weight
        user_bias = model.user_bias.weight.squeeze()
        item_bias = model.item_bias.weight.squeeze()

        # Precompute all scores? Not for memory; loop per user.
        recalls = []
        precisions = []

        # Only evaluate users who have test interactions
        test_coo = test_mat.tocoo()
        # Build dict: user -> set of test items
        test_user_items = defaultdict(set)
        for u, i, v in zip(test_coo.row, test_coo.col, test_coo.data):
            if v > 0:
                test_user_items[u].add(i)

        for u, test_items in test_user_items.items():
            # Get user vector
            u_vec = user_emb[u].unsqueeze(0)     # (1, dim)
            scores = (u_vec @ item_emb.T).squeeze() + user_bias[u] + item_bias
            # Mask training items (already seen)
            train_items = train_mat[u].indices
            scores[train_items] = -1e10
            # Top‑k
            _, topk = torch.topk(scores, k)
            topk = topk.cpu().numpy()
            # Compute recall and precision for this user
            hits = len(set(topk) & test_items)
            recall = hits / len(test_items) if test_items else 0.0
            precision = hits / k
            recalls.append(recall)
            precisions.append(precision)

        return np.mean(recalls) if recalls else 0.0, np.mean(precisions) if precisions else 0.0

# ---------------------------------------------------------------------
# 9. Main
# ---------------------------------------------------------------------
def main():
    # ---------- Load your data ----------
    # df_tracks = pd.read_csv('tracks.csv')
    # ------------------------------------

    print("Parsing reactions...")
    df_int = parse_track_reactions(df_tracks)
    print(f"Total raw interactions: {len(df_int)}")

    print("Computing poster affinities...")
    df_aff = compute_poster_affinity(df_int)

    print("Building debiased matrix and item features...")
    mat, user_enc, item_enc, df_pos, item_features = build_debiased_matrix(
        df_int, df_aff, df_tracks, alpha=ALPHA
    )
    print(f"After debiasing: {len(df_pos)} positive interactions")
    print(f"Users: {mat.shape[0]}, Items: {mat.shape[1]}, Interactions: {mat.nnz}")

    n_users, n_items = mat.shape
    item_feat_dim = item_features.shape[1] if item_features is not None else None
    print(f"Artist feature dimension: {item_feat_dim}")

    # Train/test split (random interaction split)
    coo = mat.tocoo()
    np.random.seed(42)
    mask = np.random.rand(len(coo.data)) < (1 - TEST_PERCENT)
    train_mat = csr_matrix((coo.data[mask], (coo.row[mask], coo.col[mask])), shape=mat.shape)
    test_mat  = csr_matrix((coo.data[~mask], (coo.row[~mask], coo.col[~mask])), shape=mat.shape)

    # Prepare DataLoader
    train_rows = train_mat.tocoo().row
    train_cols = train_mat.tocoo().col
    train_df = pd.DataFrame({'user_idx': train_rows, 'item_idx': train_cols})
    dataset = BPRDataset(train_df, n_items, num_negative=NUM_NEGATIVE)
    dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)

    # Convert item features to dense tensor (or sparse) for fast lookup
    if item_features is not None:
        item_features_tensor = torch.FloatTensor(item_features.toarray()).to(DEVICE)
    else:
        item_features_tensor = None

    # Model
    model = HybridMF(n_users, n_items, EMBEDDING_DIM, item_feat_dim).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=LR)
    scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.5)

    # Training loop
    for epoch in range(1, EPOCHS + 1):
        loss = train_epoch(model, dataloader, optimizer, item_features_tensor, DEVICE)
        scheduler.step()
        print(f"Epoch {epoch:02d}  loss={loss:.4f}  lr={scheduler.get_last_lr()[0]:.6f}")
        if epoch % 3 == 0 or epoch == EPOCHS:
            rec, prec = evaluate_per_user(model, test_mat, train_mat, k=10)
            print(f"   Per‑user Test Recall@10 = {rec:.4f}  Precision@10 = {prec:.4f}")

    # Save model
    torch.save({
        'model_state_dict': model.state_dict(),
        'user_enc': user_enc,
        'item_enc': item_enc,
    }, 'hybrid_mf_model.pth')
    print("Model saved to hybrid_mf_model.pth")

if __name__ == "__main__":
    main()
```

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from sklearn.preprocessing import LabelEncoder
from scipy.sparse import csr_matrix
import random
from collections import defaultdict

# ---------------------------------------------------------------------
# 0. Configuration
# ---------------------------------------------------------------------
FIXED_LIKE_ENERGY = 1
FIXED_DISLIKE_ENERGY = -2
ALPHA = 0.5                     # debiasing strength
EMBEDDING_DIM = 64
NUM_LAYERS = 3                  # LightGCN layers
BATCH_SIZE = 1024
EPOCHS = 50
LR = 0.001
WEIGHT_DECAY = 1e-5             # L2 regularization
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
TEST_PERCENT = 0.2
NUM_NEGATIVE = 8

# ---------------------------------------------------------------------
# 1. Parse reactions (same)
# ---------------------------------------------------------------------
def parse_track_reactions(df_tracks):
    interactions = []
    for _, row in df_tracks.iterrows():
        track_id = row['track_id']
        sender_ids = str(row['sender_chat_ids'])
        sender_list = [int(x.strip()) for x in sender_ids.split(',') if x.strip()]
        sender = sender_list[0] if len(sender_list) == 1 else -1

        if pd.notna(row['likes']):
            for uid in str(row['likes']).split(','):
                uid = int(uid.strip())
                interactions.append((uid, track_id, FIXED_LIKE_ENERGY, sender))
        if pd.notna(row['dislikes']):
            for uid in str(row['dislikes']).split(','):
                uid = int(uid.strip())
                interactions.append((uid, track_id, FIXED_DISLIKE_ENERGY, sender))
        if pd.notna(row['reactions']):
            for token in str(row['reactions']).split(','):
                token = token.strip()
                if ':' not in token:
                    continue
                uid_str, energy_str = token.split(':')
                uid = int(uid_str)
                energy = int(energy_str)
                interactions.append((uid, track_id, energy, sender))
    return pd.DataFrame(interactions, columns=['user_id','track_id','energy','sender_id'])

# ---------------------------------------------------------------------
# 2. Poster affinity and debiased matrix (unchanged except encoder fitting)
# ---------------------------------------------------------------------
def compute_poster_affinity(df_int):
    pos = df_int[df_int['energy'] > 0]
    if pos.empty:
        return pd.DataFrame(columns=['user_id','sender_id','affinity'])
    aff = pos.groupby(['user_id','sender_id'])['energy'].mean().reset_index()
    aff.rename(columns={'energy':'affinity'}, inplace=True)
    return aff

def build_debiased_matrix(df_int, df_aff, df_tracks, alpha=ALPHA):
    df = df_int.merge(df_aff, on=['user_id','sender_id'], how='left')
    df['affinity'] = df['affinity'].fillna(0)
    df['weight'] = (df['energy'] - alpha * df['affinity']).clip(lower=0)
    df_pos = df[df['weight'] > 0].copy()
    df_neg = df[df['energy'] < 0].copy()

    # Fit encoders on all users/items (positive + negative) to avoid unseen labels
    all_user_ids = pd.concat([df_pos['user_id'], df_neg['user_id']]).unique()
    all_item_ids = pd.concat([df_pos['track_id'], df_neg['track_id']]).unique()
    user_enc = LabelEncoder()
    user_enc.fit(all_user_ids)
    item_enc = LabelEncoder()
    item_enc.fit(all_item_ids)

    df_pos['user_idx'] = user_enc.transform(df_pos['user_id'])
    df_pos['item_idx'] = item_enc.transform(df_pos['track_id'])
    n_users = len(user_enc.classes_)
    n_items = len(item_enc.classes_)

    mat = csr_matrix((df_pos['weight'], (df_pos['user_idx'], df_pos['item_idx'])),
                     shape=(n_users, n_items))

    # Build dislike pairs (not used in this stable version, but keep for later)
    dislike_pairs = defaultdict(set)
    valid_neg = df_neg['user_id'].isin(user_enc.classes_) & df_neg['track_id'].isin(item_enc.classes_)
    df_neg = df_neg[valid_neg]
    if not df_neg.empty:
        df_neg['user_idx'] = user_enc.transform(df_neg['user_id'])
        df_neg['item_idx'] = item_enc.transform(df_neg['track_id'])
        for u, i in zip(df_neg['user_idx'], df_neg['item_idx']):
            dislike_pairs[u].add(i)

    return mat, user_enc, item_enc, df_pos, dislike_pairs

# ---------------------------------------------------------------------
# 3. Item & user feature construction (same as before)
# ---------------------------------------------------------------------
def build_item_features(df_tracks, item_enc):
    # artist one-hot
    artist_enc = LabelEncoder()
    artist_enc.fit(df_tracks['artist_names'])
    n_items = len(item_enc.classes_)
    n_artists = len(artist_enc.classes_)
    artist_mat = np.zeros((n_items, n_artists), dtype=np.float32)
    for _, row in df_tracks.iterrows():
        tid = row['track_id']
        if tid in item_enc.classes_:
            idx = item_enc.transform([tid])[0]
            artist_idx = artist_enc.transform([row['artist_names']])[0]
            artist_mat[idx, artist_idx] = 1.0

    # emoji energy counts
    all_energies = set()
    track_energy_counts = defaultdict(lambda: defaultdict(int))
    for _, row in df_tracks.iterrows():
        tid = row['track_id']
        if pd.notna(row['likes']):
            for _ in str(row['likes']).split(','):
                track_energy_counts[tid][FIXED_LIKE_ENERGY] += 1
                all_energies.add(FIXED_LIKE_ENERGY)
        if pd.notna(row['dislikes']):
            for _ in str(row['dislikes']).split(','):
                track_energy_counts[tid][FIXED_DISLIKE_ENERGY] += 1
                all_energies.add(FIXED_DISLIKE_ENERGY)
        if pd.notna(row['reactions']):
            for token in str(row['reactions']).split(','):
                token = token.strip()
                if ':' not in token: continue
                _, energy_str = token.split(':')
                energy = int(energy_str)
                track_energy_counts[tid][energy] += 1
                all_energies.add(energy)

    energy_list = sorted(all_energies)
    energy_to_idx = {e:i for i,e in enumerate(energy_list)}
    n_energies = len(energy_list)
    emoji_mat = np.zeros((n_items, n_energies), dtype=np.float32)
    for tid, counts in track_energy_counts.items():
        if tid in item_enc.classes_:
            idx = item_enc.transform([tid])[0]
            for energy, cnt in counts.items():
                emoji_mat[idx, energy_to_idx[energy]] = cnt

    # Concatenate artist + emoji
    item_features = np.concatenate([artist_mat, emoji_mat], axis=1)
    return torch.FloatTensor(item_features)

def build_user_features(df_users, user_enc):
    if df_users is None:
        return None
    n_users = len(user_enc.classes_)
    feature_cols = ['num_likes','num_dislikes','num_reactions',
                    'num_likes_received','num_dislikes_received','num_reactions_received',
                    'num_liked_users','num_disliked_users','num_reacted_users',
                    'num_users_liked','num_users_disliked','num_users_reacted']
    user_feat = np.zeros((n_users, len(feature_cols)), dtype=np.float32)
    for i, uid in enumerate(user_enc.classes_):
        row = df_users[df_users['user_id'] == uid]
        if not row.empty:
            vals = [row[c].values[0] if c in row.columns else 0 for c in feature_cols]
            user_feat[i] = np.array(vals, dtype=np.float32)
    return torch.FloatTensor(user_feat)

# ---------------------------------------------------------------------
# 4. LightGCN (simplified, stable)
# ---------------------------------------------------------------------
class LightGCN(nn.Module):
    def __init__(self, n_users, n_items, emb_dim, num_layers,
                 user_feat_dim=None, item_feat_dim=None):
        super().__init__()
        self.n_users = n_users
        self.n_items = n_items
        self.emb_dim = emb_dim
        self.num_layers = num_layers

        # Feature projections (optional)
        if user_feat_dim:
            self.user_feat_proj = nn.Linear(user_feat_dim, emb_dim, bias=False)
        else:
            self.user_feat_proj = None
        if item_feat_dim:
            self.item_feat_proj = nn.Linear(item_feat_dim, emb_dim, bias=False)
        else:
            self.item_feat_proj = None

        # ID embeddings
        self.user_emb = nn.Embedding(n_users, emb_dim)
        self.item_emb = nn.Embedding(n_items, emb_dim)
        nn.init.xavier_uniform_(self.user_emb.weight)
        nn.init.xavier_uniform_(self.item_emb.weight)

    def compute_initial_embeddings(self, user_features=None, item_features=None):
        u = self.user_emb.weight
        i = self.item_emb.weight
        if self.user_feat_proj is not None and user_features is not None:
            u = u + self.user_feat_proj(user_features.to(DEVICE))
        if self.item_feat_proj is not None and item_features is not None:
            i = i + self.item_feat_proj(item_features.to(DEVICE))
        return u, i

    def forward(self, adj, user_features=None, item_features=None):
        ego_u, ego_i = self.compute_initial_embeddings(user_features, item_features)
        all_emb = torch.cat([ego_u, ego_i], dim=0)      # (N, dim)
        emb_list = [all_emb]
        for _ in range(self.num_layers):
            all_emb = torch.sparse.mm(adj, all_emb)      # graph convolution
            emb_list.append(all_emb)
        final_emb = torch.stack(emb_list, dim=1).mean(dim=1)  # mean over layers
        user_emb, item_emb = torch.split(final_emb, [self.n_users, self.n_items])
        return user_emb, item_emb

# ---------------------------------------------------------------------
# 5. BPR Dataset (simple, random negatives)
# ---------------------------------------------------------------------
class BPRDataset(Dataset):
    def __init__(self, interactions_df, n_items, num_negative=8):
        self.users = torch.LongTensor(interactions_df['user_idx'].values)
        self.pos_items = torch.LongTensor(interactions_df['item_idx'].values)
        self.n_items = n_items
        self.num_negative = num_negative

    def __len__(self):
        return len(self.users)

    def __getitem__(self, idx):
        user = self.users[idx]
        pos = self.pos_items[idx]
        negs = []
        while len(negs) < self.num_negative:
            neg = random.randint(0, self.n_items - 1)
            if neg != pos and neg not in negs:
                negs.append(neg)
        return user, pos, torch.LongTensor(negs)

# ---------------------------------------------------------------------
# 6. BPR loss (pure, no dislike penalty)
# ---------------------------------------------------------------------
def bpr_loss(user_emb, item_emb, user, pos_item, neg_items):
    u = user_emb[user]                          # (batch, dim)
    pos = item_emb[pos_item]                    # (batch, dim)
    neg = item_emb[neg_items]                   # (batch, num_neg, dim)
    pos_score = (u * pos).sum(dim=-1)           # (batch,)
    neg_scores = (u.unsqueeze(1) * neg).sum(dim=-1)  # (batch, num_neg)
    diff = pos_score.unsqueeze(1) - neg_scores
    return -torch.log(torch.sigmoid(diff) + 1e-10).mean()

# ---------------------------------------------------------------------
# 7. Evaluation (unchanged)
# ---------------------------------------------------------------------
def evaluate_per_user(user_emb, item_emb, test_mat, train_mat, k=10):
    with torch.no_grad():
        recalls, precisions = [], []
        test_coo = test_mat.tocoo()
        test_user_items = defaultdict(set)
        for u, i, v in zip(test_coo.row, test_coo.col, test_coo.data):
            if v > 0:
                test_user_items[u].add(i)

        for u, test_items in test_user_items.items():
            scores = (user_emb[u].unsqueeze(0) @ item_emb.T).squeeze()
            train_items = train_mat[u].indices
            scores[train_items] = -1e10
            _, topk = torch.topk(scores, k)
            topk = topk.cpu().numpy()
            hits = len(set(topk) & test_items)
            recall = hits / len(test_items) if test_items else 0.0
            precision = hits / k
            recalls.append(recall)
            precisions.append(precision)
        return np.mean(recalls) if recalls else 0.0, np.mean(precisions) if precisions else 0.0

# ---------------------------------------------------------------------
# 8. Build normalized adjacency (fixed)
# ---------------------------------------------------------------------
def build_adjacency(mat):
    n_users, n_items = mat.shape
    N = n_users + n_items
    coo = mat.tocoo()
    row = np.concatenate([coo.row, coo.col + n_users])
    col = np.concatenate([coo.col + n_users, coo.row])
    data = np.ones(len(row), dtype=np.float32)
    adj = csr_matrix((data, (row, col)), shape=(N, N))
    # Symmetric normalization D^{-1/2} A D^{-1/2}
    deg = np.array(adj.sum(axis=1)).flatten()
    deg_inv_sqrt = np.power(deg, -0.5, where=deg>0)
    deg_inv_sqrt[np.isinf(deg_inv_sqrt)] = 0.0
    D = csr_matrix((deg_inv_sqrt, (np.arange(N), np.arange(N))), shape=(N, N))
    adj_norm = D * adj * D
    coo = adj_norm.tocoo()
    indices = torch.LongTensor(np.vstack([coo.row, coo.col]))
    values = torch.FloatTensor(coo.data)
    adj_tensor = torch.sparse_coo_tensor(indices, values, torch.Size([N, N]))
    return adj_tensor

# ---------------------------------------------------------------------
# 9. Main training
# ---------------------------------------------------------------------
def main():
    # Load your data: df_tracks, df_users
    # df_tracks = ...
    # df_users = ...

    print("Parsing reactions...")
    df_int = parse_track_reactions(df_tracks)
    print(f"Total raw interactions: {len(df_int)}")
    df_aff = compute_poster_affinity(df_int)

    print("Building debiased matrix...")
    mat, user_enc, item_enc, df_pos, dislike_pairs = build_debiased_matrix(
        df_int, df_aff, df_tracks, alpha=ALPHA
    )
    print(f"Positive interactions: {mat.nnz}, Users: {mat.shape[0]}, Items: {mat.shape[1]}")

    # Build features
    print("Building item features...")
    item_features = build_item_features(df_tracks, item_enc).to(DEVICE)
    item_feat_dim = item_features.shape[1]
    print("Building user features...")
    user_features = build_user_features(df_users, user_enc)
    if user_features is not None:
        user_features = user_features.to(DEVICE)
        user_feat_dim = user_features.shape[1]
    else:
        user_feat_dim = None

    # Train/test split
    coo = mat.tocoo()
    np.random.seed(42)
    mask = np.random.rand(len(coo.data)) < (1 - TEST_PERCENT)
    train_mat = csr_matrix((coo.data[mask], (coo.row[mask], coo.col[mask])), shape=mat.shape)
    test_mat = csr_matrix((coo.data[~mask], (coo.row[~mask], coo.col[~mask])), shape=mat.shape)

    # Build adjacency (binary) from training matrix
    train_binary = train_mat.copy()
    train_binary.data = np.ones_like(train_binary.data)
    adj = build_adjacency(train_binary).to(DEVICE)

    # DataLoader
    train_rows = train_mat.tocoo().row
    train_cols = train_mat.tocoo().col
    train_df = pd.DataFrame({'user_idx': train_rows, 'item_idx': train_cols})
    dataset = BPRDataset(train_df, mat.shape[1], num_negative=NUM_NEGATIVE)
    dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)

    # Model
    model = LightGCN(mat.shape[0], mat.shape[1], EMBEDDING_DIM, NUM_LAYERS,
                     user_feat_dim=user_feat_dim, item_feat_dim=item_feat_dim).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS, eta_min=1e-6)

    # Training loop
    for epoch in range(1, EPOCHS+1):
        model.train()
        total_loss = 0
        for user, pos, negs in dataloader:
            user, pos, negs = user.to(DEVICE), pos.to(DEVICE), negs.to(DEVICE)
            
            # ---- compute final embeddings inside the loop (fresh graph each batch) ----
            user_emb, item_emb = model(adj, user_features, item_features)
            
            optimizer.zero_grad()
            loss = bpr_loss(user_emb, item_emb, user, pos, negs)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
        scheduler.step()

        if epoch % 3 == 0 or epoch == EPOCHS:
            # Evaluate with a fresh forward pass (no gradients needed)
            model.eval()
            with torch.no_grad():
                user_emb_eval, item_emb_eval = model(adj, user_features, item_features)
            rec, prec = evaluate_per_user(user_emb_eval, item_emb_eval, test_mat, train_mat, k=10)
            model.train()
            print(f"Epoch {epoch:02d} loss={total_loss/len(dataloader):.4f}  recall@10={rec:.4f} precision@10={prec:.4f}")

    torch.save({
        'model_state': model.state_dict(),
        'user_enc': user_enc,
        'item_enc': item_enc,
    }, 'lightgcn_model.pth')
    print("Model saved.")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import defaultdict

# -------------------------------------------------------------------
# 1. Parse senders and reactions into usable format
# -------------------------------------------------------------------
def parse_sender_list(sender_str):
    """Return list of sender user IDs (ints), skip '0'."""
    if pd.isna(sender_str):
        return []
    return [int(x.strip()) for x in str(sender_str).split(',') if x.strip() != '0']

def parse_reactors(row):
    """Return list of user IDs who reacted to a track (likes + reactions)."""
    reactors = []
    if pd.notna(row['likes']):
        for uid in str(row['likes']).split(','):
            uid = uid.strip()
            if uid:
                reactors.append(int(uid))
    if pd.notna(row['reactions']):
        for token in str(row['reactions']).split(','):
            token = token.strip()
            if ':' in token:
                uid = int(token.split(':')[0])
                reactors.append(uid)
    return reactors

# -------------------------------------------------------------------
# 2. Sender influence baseline
# -------------------------------------------------------------------
print("Building sender influence baseline...")

# Accumulate total reactions per sender across all their tracks
sender_total_reactions = defaultdict(int)
sender_track_count = defaultdict(int)

for _, row in df_tracks.iterrows():
    senders = parse_sender_list(row['sender_chat_ids'])
    reactions = row['total_reactions']
    for s in senders:
        sender_total_reactions[s] += reactions
        sender_track_count[s] += 1

# Average reactions per track for each sender
sender_mean = {}
for s in sender_track_count:
    sender_mean[s] = sender_total_reactions[s] / sender_track_count[s]

# For tracks with unknown sender (only "0"), use global mean
global_mean = df_tracks['total_reactions'].mean()

def get_expected_reactions(row):
    senders = parse_sender_list(row['sender_chat_ids'])
    if not senders:
        return global_mean
    # Use the first sender's average (or could average across senders)
    return sender_mean.get(senders[0], global_mean)

# -------------------------------------------------------------------
# 3. Compute residuals and flag suspicious tracks
# -------------------------------------------------------------------
print("Computing residuals...")
df_tracks['expected_reactions'] = df_tracks.apply(get_expected_reactions, axis=1)
df_tracks['residual'] = df_tracks['total_reactions'] - df_tracks['expected_reactions']

# Overall residual standard deviation (robust to skew)
res_std = df_tracks['residual'].std()
df_tracks['residual_z'] = df_tracks['residual'] / res_std

# Flag tracks where actual reactions are much higher than expected
THRESHOLD_Z = 2.0   # can adjust
flagged_by_residual = df_tracks[df_tracks['residual_z'] > THRESHOLD_Z]

print(f"Flagged by residual (> {THRESHOLD_Z} std above expected): {len(flagged_by_residual)} tracks")

# -------------------------------------------------------------------
# 4. Reactor-sender affinity check
# -------------------------------------------------------------------
print("Building user-sender affinity matrix...")

# For each (user, sender) pair, count interactions
user_sender_count = defaultdict(int)
user_sender_energy = defaultdict(int)   # optional

# Loop through all tracks to get user reactions and track senders
for _, row in df_tracks.iterrows():
    senders = parse_sender_list(row['sender_chat_ids'])
    if not senders:
        continue
    primary_sender = senders[0]  # simplify: use first sender
    reactors = parse_reactors(row)
    for u in reactors:
        user_sender_count[(u, primary_sender)] += 1
        # could also sum energy; skip for now

# Compute per-track average affinity of its reactors to its sender
def track_affinity(row):
    senders = parse_sender_list(row['sender_chat_ids'])
    if not senders:
        return np.nan
    primary_sender = senders[0]
    reactors = parse_reactors(row)
    if not reactors:
        return np.nan
    total_aff = 0
    for u in reactors:
        total_aff += user_sender_count.get((u, primary_sender), 0)
    return total_aff / len(reactors)

df_tracks['reactor_affinity'] = df_tracks.apply(track_affinity, axis=1)

# Baseline: average affinity across all tracks (or reactions)
avg_affinity = df_tracks['reactor_affinity'].mean()
aff_std = df_tracks['reactor_affinity'].std()
df_tracks['affinity_z'] = (df_tracks['reactor_affinity'] - avg_affinity) / aff_std

flagged_by_affinity = df_tracks[df_tracks['affinity_z'] > 2.0]

print(f"Flagged by high reactor affinity (> 2 std): {len(flagged_by_affinity)} tracks")

# -------------------------------------------------------------------
# 5. Tracks flagged by BOTH criteria are strong candidates
# -------------------------------------------------------------------
both_flags = set(flagged_by_residual['track_id']) & set(flagged_by_affinity['track_id'])
print(f"Tracks flagged by both criteria: {len(both_flags)}")

# Show some examples
if both_flags:
    display_cols = ['track_id', 'track_name', 'artist_names', 'total_reactions',
                    'expected_reactions', 'num_senders', 'residual_z', 'affinity_z']
    print(df_tracks[df_tracks['track_id'].isin(both_flags)][display_cols].head(20))

# -------------------------------------------------------------------
# 6. Visualization
# -------------------------------------------------------------------
plt.figure(figsize=(10, 6))
# Plot all tracks in gray
plt.scatter(df_tracks['expected_reactions'], df_tracks['total_reactions'],
            alpha=0.3, label='Normal')
# Highlight flagged tracks
flagged = df_tracks[df_tracks['track_id'].isin(both_flags)]
plt.scatter(flagged['expected_reactions'], flagged['total_reactions'],
            color='red', alpha=0.7, label='Socially suspect (both flags)')
# Diagonal reference
mx = max(df_tracks['expected_reactions'].max(), df_tracks['total_reactions'].max())
plt.plot([0, mx], [0, mx], 'k--', alpha=0.5, label='Expected = Actual')
plt.xlabel('Expected reactions (based on sender average)')
plt.ylabel('Actual total reactions')
plt.title('Track Reactions vs Sender Expectation')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# Distribution of residuals
plt.figure(figsize=(8, 4))
plt.hist(df_tracks['residual_z'], bins=50, color='steelblue', edgecolor='white')
plt.axvline(2.0, color='red', linestyle='--', label='Flag threshold (z=2)')
plt.xlabel('Residual z‑score')
plt.ylabel('Number of tracks')
plt.title('Residual Distribution (Actual – Expected)')
plt.legend()
plt.show()

print("\nDone.")
```

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from sklearn.preprocessing import LabelEncoder
from scipy.sparse import csr_matrix
from collections import defaultdict
import random

# -------------------------------------------------------------------
# CONFIG
# -------------------------------------------------------------------
FIXED_LIKE_ENERGY = 1
FIXED_DISLIKE_ENERGY = -2
ALPHA = 0.0                     # no debiasing for this test
EMBEDDING_DIM = 64
BATCH_SIZE = 2048
EPOCHS = 20                     # quick training
LR = 0.001
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
TEST_PERCENT = 0.2

# -------------------------------------------------------------------
# 1. Parse reactions (same as before)
# -------------------------------------------------------------------
def parse_track_reactions(df_tracks):
    interactions = []
    for _, row in df_tracks.iterrows():
        tid = row['track_id']
        if pd.notna(row['likes']):
            for uid in str(row['likes']).split(','):
                interactions.append((int(uid.strip()), tid, FIXED_LIKE_ENERGY))
        if pd.notna(row['reactions']):
            for token in str(row['reactions']).split(','):
                token = token.strip()
                if ':' not in token: continue
                uid_str, energy_str = token.split(':')
                interactions.append((int(uid_str), tid, int(energy_str)))
    return pd.DataFrame(interactions, columns=['user_id','track_id','energy'])

# -------------------------------------------------------------------
# 2. Build positive interaction matrix (all energies > 0)
# -------------------------------------------------------------------
def build_matrix(df_int):
    pos = df_int[df_int['energy'] > 0].copy()
    user_enc = LabelEncoder()
    item_enc = LabelEncoder()
    pos['user_idx'] = user_enc.fit_transform(pos['user_id'])
    pos['item_idx'] = item_enc.fit_transform(pos['track_id'])
    mat = csr_matrix((np.ones(len(pos)), (pos['user_idx'], pos['item_idx'])),
                     shape=(len(user_enc.classes_), len(item_enc.classes_)))
    return mat, user_enc, item_enc

# -------------------------------------------------------------------
# 3. Feature builders
# -------------------------------------------------------------------
def build_artist_features(df_tracks, item_enc):
    artist_enc = LabelEncoder()
    artist_enc.fit(df_tracks['artist_names'])
    n_items = len(item_enc.classes_)
    n_artists = len(artist_enc.classes_)
    feat = np.zeros((n_items, n_artists), dtype=np.float32)
    for _, row in df_tracks.iterrows():
        tid = row['track_id']
        if tid in item_enc.classes_:
            idx = item_enc.transform([tid])[0]
            artist_idx = artist_enc.transform([row['artist_names']])[0]
            feat[idx, artist_idx] = 1.0
    return torch.FloatTensor(feat)

def build_emoji_features(df_tracks, item_enc):
    # energy level counts per track
    energy_levels = set()
    track_energy = defaultdict(lambda: defaultdict(int))
    for _, row in df_tracks.iterrows():
        tid = row['track_id']
        if pd.notna(row['likes']):
            for _ in str(row['likes']).split(','):
                track_energy[tid][FIXED_LIKE_ENERGY] += 1
                energy_levels.add(FIXED_LIKE_ENERGY)
        if pd.notna(row['reactions']):
            for token in str(row['reactions']).split(','):
                token = token.strip()
                if ':' not in token: continue
                energy = int(token.split(':')[1])
                track_energy[tid][energy] += 1
                energy_levels.add(energy)
    energy_list = sorted(energy_levels)
    energy_to_idx = {e:i for i,e in enumerate(energy_list)}
    n_items = len(item_enc.classes_)
    n_energies = len(energy_list)
    feat = np.zeros((n_items, n_energies), dtype=np.float32)
    for tid, counts in track_energy.items():
        if tid in item_enc.classes_:
            idx = item_enc.transform([tid])[0]
            for e, cnt in counts.items():
                feat[idx, energy_to_idx[e]] = cnt
    # normalize rows to unit length
    norm = np.linalg.norm(feat, axis=1, keepdims=True) + 1e-10
    feat = feat / norm
    return torch.FloatTensor(feat)

def build_user_features(df_users, user_enc):
    if df_users is None:
        return None
    n_users = len(user_enc.classes_)
    cols = ['num_likes','num_dislikes','num_reactions',
            'num_likes_received','num_dislikes_received','num_reactions_received',
            'num_users_liked','num_users_disliked','num_users_reacted']
    feat = np.zeros((n_users, len(cols)), dtype=np.float32)
    for i, uid in enumerate(user_enc.classes_):
        row = df_users[df_users['user_id'] == uid]
        if not row.empty:
            vals = [row[c].values[0] if c in row.columns else 0 for c in cols]
            feat[i] = np.array(vals, dtype=np.float32)
    # normalize
    mean = feat.mean(axis=0, keepdims=True)
    std = feat.std(axis=0, keepdims=True) + 1e-10
    feat = (feat - mean) / std
    return torch.FloatTensor(feat)

# -------------------------------------------------------------------
# 4. Model: MF with optional side features
# -------------------------------------------------------------------
class ContentMF(nn.Module):
    def __init__(self, n_users, n_items, emb_dim, user_feat_dim=0, item_feat_dim=0):
        super().__init__()
        self.user_emb = nn.Embedding(n_users, emb_dim)
        self.item_emb = nn.Embedding(n_items, emb_dim)
        nn.init.xavier_uniform_(self.user_emb.weight)
        nn.init.xavier_uniform_(self.item_emb.weight)
        self.user_proj = nn.Linear(user_feat_dim, emb_dim, bias=False) if user_feat_dim else None
        self.item_proj = nn.Linear(item_feat_dim, emb_dim, bias=False) if item_feat_dim else None

    def forward(self, user_idx, item_idx, user_feat=None, item_feat=None):
        u = self.user_emb(user_idx)
        i = self.item_emb(item_idx)
        if self.user_proj is not None and user_feat is not None:
            u = u + self.user_proj(user_feat[user_idx])
        if self.item_proj is not None and item_feat is not None:
            i = i + self.item_proj(item_feat[item_idx])
        return (u * i).sum(dim=-1)

# -------------------------------------------------------------------
# 5. BPR loss
# -------------------------------------------------------------------
def bpr_loss(model, user, pos, neg, uf=None, ift=None):
    pos_score = model(user, pos, uf, ift)
    neg_score = model(user, neg, uf, ift)
    diff = pos_score - neg_score
    return -torch.log(torch.sigmoid(diff) + 1e-10).mean()

# -------------------------------------------------------------------
# 6. Dataset
# -------------------------------------------------------------------
class BPRDataset(Dataset):
    def __init__(self, interactions_df, n_items, num_neg=1):
        self.users = torch.LongTensor(interactions_df['user_idx'].values)
        self.pos_items = torch.LongTensor(interactions_df['item_idx'].values)
        self.n_items = n_items
        self.num_neg = num_neg

    def __len__(self):
        return len(self.users)

    def __getitem__(self, idx):
        user = self.users[idx]
        pos = self.pos_items[idx]
        negs = []
        while len(negs) < self.num_neg:
            neg = random.randint(0, self.n_items - 1)
            if neg != pos:
                negs.append(neg)
        return user, pos, torch.LongTensor(negs)

# -------------------------------------------------------------------
# 7. Evaluation
# -------------------------------------------------------------------
def evaluate(model, user_emb, item_emb, test_mat, train_mat, uf=None, ift=None, k=10):
    model.eval()
    with torch.no_grad():
        # get final embeddings (with features)
        if uf is not None:
            u_emb = model.user_emb.weight + model.user_proj(uf)
        else:
            u_emb = model.user_emb.weight
        if ift is not None:
            i_emb = model.item_emb.weight + model.item_proj(ift)
        else:
            i_emb = model.item_emb.weight

        recalls, precisions = [], []
        test_coo = test_mat.tocoo()
        test_user_items = defaultdict(set)
        for u, i, v in zip(test_coo.row, test_coo.col, test_coo.data):
            if v > 0:
                test_user_items[u].add(i)

        for u, test_items in test_user_items.items():
            scores = u_emb[u].unsqueeze(0) @ i_emb.T
            train_items = train_mat[u].indices
            scores[0, train_items] = -1e10
            _, topk = torch.topk(scores, k)
            topk = topk.cpu().numpy().flatten()
            hits = len(set(topk) & test_items)
            recall = hits / len(test_items) if test_items else 0.0
            precision = hits / k
            recalls.append(recall)
            precisions.append(precision)
        return np.mean(recalls) if recalls else 0.0, np.mean(precisions) if precisions else 0.0

# -------------------------------------------------------------------
# 8. Main test loop
# -------------------------------------------------------------------
def test_feature_set(name, user_feat_tensor, item_feat_tensor, train_mat, test_mat):
    n_users, n_items = train_mat.shape
    uf_dim = user_feat_tensor.shape[1] if user_feat_tensor is not None else 0
    if_dim = item_feat_tensor.shape[1] if item_feat_tensor is not None else 0

    train_coo = train_mat.tocoo()
    train_df = pd.DataFrame({'user_idx': train_coo.row, 'item_idx': train_coo.col})
    dataset = BPRDataset(train_df, n_items, num_neg=1)
    dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True)

    model = ContentMF(n_users, n_items, EMBEDDING_DIM, uf_dim, if_dim).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=1e-5)
    scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=10, gamma=0.5)

    best_recall = 0.0
    for epoch in range(1, EPOCHS+1):
        model.train()
        for user, pos, neg in dataloader:
            user, pos, neg = user.to(DEVICE), pos.to(DEVICE), neg.squeeze(1).to(DEVICE)
            optimizer.zero_grad()
            loss = bpr_loss(model, user, pos, neg, user_feat_tensor, item_feat_tensor)
            loss.backward()
            optimizer.step()
        scheduler.step()

        # Evaluate every 5 epochs
        if epoch % 5 == 0:
            rec, prec = evaluate(model, user_feat_tensor, item_feat_tensor,
                                 test_mat, train_mat, k=10)
            if rec > best_recall:
                best_recall = rec
    print(f"{name:30s} | Best Recall@10 = {best_recall:.4f}")
    return best_recall

# -------------------------------------------------------------------
# 9. Run
# -------------------------------------------------------------------
def main():
    # Load data (adjust to your actual loading)
    # df_tracks = ...
    # df_users  = ...

    print("Parsing reactions...")
    df_int = parse_track_reactions(df_tracks)
    print(f"Total interactions: {len(df_int)}")
    mat, user_enc, item_enc = build_matrix(df_int)
    print(f"Matrix: {mat.shape[0]} users x {mat.shape[1]} items, {mat.nnz} nnz")

    # Train/test split
    coo = mat.tocoo()
    np.random.seed(42)
    mask = np.random.rand(len(coo.data)) < (1 - TEST_PERCENT)
    train_mat = csr_matrix((coo.data[mask], (coo.row[mask], coo.col[mask])), shape=mat.shape)
    test_mat  = csr_matrix((coo.data[~mask], (coo.row[~mask], coo.col[~mask])), shape=mat.shape)

    # Precompute feature sets
    print("\nBuilding feature sets...")
    artist_feat = build_artist_features(df_tracks, item_enc)
    emoji_feat  = build_emoji_features(df_tracks, item_enc)
    artist_emoji_feat = torch.cat([artist_feat, emoji_feat], dim=1)
    user_feat   = build_user_features(df_users, user_enc)

    # Move to device
    artist_feat = artist_feat.to(DEVICE) if artist_feat is not None else None
    emoji_feat = emoji_feat.to(DEVICE) if emoji_feat is not None else None
    artist_emoji_feat = artist_emoji_feat.to(DEVICE)
    user_feat = user_feat.to(DEVICE) if user_feat is not None else None

    print("\n=== Feature Engineering Comparison ===")
    results = {}
    results['Baseline (no features)'] = test_feature_set('Baseline', None, None, train_mat, test_mat)
    results['Artist only']            = test_feature_set('Artist only', None, artist_feat, train_mat, test_mat)
    results['Emoji profile only']     = test_feature_set('Emoji profile', None, emoji_feat, train_mat, test_mat)
    results['Artist + Emoji']         = test_feature_set('Artist + Emoji', None, artist_emoji_feat, train_mat, test_mat)
    results['User activity']          = test_feature_set('User activity', user_feat, None, train_mat, test_mat)
    results['Artist + User']          = test_feature_set('Artist + User', user_feat, artist_feat, train_mat, test_mat)
    results['All features']           = test_feature_set('All features', user_feat, artist_emoji_feat, train_mat, test_mat)

    print("\nSummary:")
    for k, v in results.items():
        print(f"  {k:25s} {v:.4f}")

if __name__ == "__main__":
    main()
```

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Dataset
from scipy.sparse import csr_matrix
from sklearn.preprocessing import LabelEncoder
from collections import defaultdict
import random

SEED = 43
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
# Optionally enforce deterministic behaviour (slower)
# torch.backends.cudnn.deterministic = True
# torch.backends.cudnn.benchmark = False

# -------------------------------------------------------------------
# Configuration
# -------------------------------------------------------------------
FIXED_LIKE_ENERGY = 1
TEST_PERCENT = 0.2
EMBEDDING_DIM = 200
BATCH_SIZE = 64
EPOCHS = 200
LR = 0.001
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# -------------------------------------------------------------------
# 1. Primary artist helper
# -------------------------------------------------------------------
def primary_artist(artist_names_str):
    if pd.isna(artist_names_str):
        return "unknown"
    return str(artist_names_str).split(',')[0].strip()

# -------------------------------------------------------------------
# 2. Build user-artist matrix (binary)
# -------------------------------------------------------------------
def build_user_artist_matrix(df_tracks, df_users):
    track_to_artist = {}
    for _, row in df_tracks.iterrows():
        track_to_artist[row['track_id']] = primary_artist(row['artist_names'])

    interactions = set()

    # From track reactions
    for _, row in df_tracks.iterrows():
        artist = track_to_artist[row['track_id']]
        if pd.notna(row['likes']):
            for uid in str(row['likes']).split(','):
                interactions.add((int(uid.strip()), artist))
        if pd.notna(row['reactions']):
            for token in str(row['reactions']).split(','):
                token = token.strip()
                if ':' not in token:
                    continue
                uid_str, energy_str = token.split(':')
                if int(energy_str) > 0:
                    interactions.add((int(uid_str), artist))

    # From uploads (if df_users provided)
    if df_users is not None:
        for _, row in df_users.iterrows():
            uid = row['user_id']
            if pd.isna(row['uploaded_tracks']):
                continue
            for tid in str(row['uploaded_tracks']).split(','):
                tid = int(tid.strip())
                artist = track_to_artist.get(tid, None)
                if artist:
                    interactions.add((uid, artist))

    # Encode
    users, artists = zip(*interactions)
    user_enc = LabelEncoder()
    artist_enc = LabelEncoder()
    user_idx = user_enc.fit_transform(users)
    artist_idx = artist_enc.fit_transform(artists)

    mat = csr_matrix((np.ones(len(interactions), dtype=np.float32),
                      (user_idx, artist_idx)),
                     shape=(len(user_enc.classes_), len(artist_enc.classes_)))

    # Mapping user_id -> matrix index
    user_id_to_idx = {uid: i for i, uid in enumerate(user_enc.classes_)}
    return mat, user_id_to_idx, artist_enc, track_to_artist

# -------------------------------------------------------------------
# 3. Build track-level train/test split
# -------------------------------------------------------------------
def build_track_split(df_tracks, test_percent):
    interactions = []
    for _, row in df_tracks.iterrows():
        tid = row['track_id']
        if pd.notna(row['likes']):
            for uid in str(row['likes']).split(','):
                interactions.append((int(uid.strip()), tid))
        if pd.notna(row['reactions']):
            for token in str(row['reactions']).split(','):
                token = token.strip()
                if ':' not in token:
                    continue
                uid_str, energy_str = token.split(':')
                if int(energy_str) > 0:
                    interactions.append((int(uid_str), tid))

    user_enc = LabelEncoder()
    item_enc = LabelEncoder()
    u_idx = user_enc.fit_transform([x[0] for x in interactions])
    i_idx = item_enc.fit_transform([x[1] for x in interactions])

    mat = csr_matrix((np.ones(len(interactions)), (u_idx, i_idx)),
                     shape=(len(user_enc.classes_), len(item_enc.classes_)))

    coo = mat.tocoo()
    np.random.seed(42)
    mask = np.random.rand(len(coo.data)) < (1 - test_percent)
    train = csr_matrix((coo.data[mask], (coo.row[mask], coo.col[mask])), shape=mat.shape)
    test = csr_matrix((coo.data[~mask], (coo.row[~mask], coo.col[~mask])), shape=mat.shape)

    # test tracks per user_id
    test_user_tracks = defaultdict(set)
    test_coo = test.tocoo()
    for u, i, v in zip(test_coo.row, test_coo.col, test_coo.data):
        if v > 0:
            uid = user_enc.classes_[u]
            tid = item_enc.classes_[i]
            test_user_tracks[uid].add(tid)

    # track popularity
    track_pop = np.array(mat.sum(axis=0)).flatten()
    # track_id -> internal index
    track_id_to_idx = {tid: i for i, tid in enumerate(item_enc.classes_)}
    # seen tracks per user_id
    seen_tracks = defaultdict(set)
    train_coo = train.tocoo()
    for u, i in zip(train_coo.row, train_coo.col):
        uid = user_enc.classes_[u]
        tid = item_enc.classes_[i]
        seen_tracks[uid].add(tid)

    return test_user_tracks, seen_tracks, track_pop, track_id_to_idx

# -------------------------------------------------------------------
# 4. BPR dataset for user-artist matrix
# -------------------------------------------------------------------
class BPRDataset(Dataset):
    def __init__(self, mat, num_neg=1):
        coo = mat.tocoo()
        self.users = torch.LongTensor(coo.row)
        self.pos_items = torch.LongTensor(coo.col)
        self.n_items = mat.shape[1]
        self.num_neg = num_neg

    def __len__(self):
        return len(self.users)

    def __getitem__(self, idx):
        user = self.users[idx]
        pos = self.pos_items[idx]
        negs = []
        while len(negs) < self.num_neg:
            neg = random.randint(0, self.n_items - 1)
            if neg != pos:
                negs.append(neg)
        return user, pos, torch.LongTensor(negs)

# -------------------------------------------------------------------
# 5. Matrix Factorisation model
# -------------------------------------------------------------------
class MF(nn.Module):
    def __init__(self, n_users, n_items, emb_dim):
        super().__init__()
        self.user_emb = nn.Embedding(n_users, emb_dim)
        self.item_emb = nn.Embedding(n_items, emb_dim)
        nn.init.xavier_uniform_(self.user_emb.weight)
        nn.init.xavier_uniform_(self.item_emb.weight)

    def forward(self, user, item):
        u = self.user_emb(user)
        i = self.item_emb(item)
        return (u * i).sum(dim=-1)

# -------------------------------------------------------------------
# 6. BPR loss
# -------------------------------------------------------------------
def bpr_loss(model, user, pos, neg):
    pos_score = model(user, pos)
    neg_score = model(user, neg)
    diff = pos_score - neg_score
    return -torch.log(torch.sigmoid(diff) + 1e-10).mean()

# -------------------------------------------------------------------
# 7. Train user-artist model
# -------------------------------------------------------------------
def train_user_artist_model(mat, epochs=30):
    n_users, n_items = mat.shape
    dataset = BPRDataset(mat)
    dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True)

    model = MF(n_users, n_items, EMBEDDING_DIM).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=1e-5)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS, eta_min=1e-6)

    for epoch in range(1, epochs+1):
        model.train()
        total_loss = 0
        for user, pos, negs in dataloader:
            user, pos, negs = user.to(DEVICE), pos.to(DEVICE), negs.squeeze(1).to(DEVICE)
            optimizer.zero_grad()
            loss = bpr_loss(model, user, pos, negs)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()
        scheduler.step()
        if epoch % 5 == 0:
            print(f"  epoch {epoch:02d}, loss={total_loss/len(dataloader):.4f}")

    # return embeddings as numpy
    model.eval()
    with torch.no_grad():
        user_emb = model.user_emb.weight.cpu().numpy()
        item_emb = model.item_emb.weight.cpu().numpy()
    return user_emb, item_emb

# -------------------------------------------------------------------
# 8. Evaluation: track recall from artist recommendations
# -------------------------------------------------------------------
def evaluate_track_recall_and_precision(user_emb, item_emb, user_id_to_idx, artist_enc,
                                        track_to_artist, test_user_tracks, seen_tracks,
                                        track_pop, track_id_to_idx, top_k=10,
                                        n_artists=5, tracks_per_artist=5):
    artist_to_tracks = defaultdict(list)
    for tid, artist_name in track_to_artist.items():
        if artist_name in artist_enc.classes_ and tid in track_id_to_idx:
            aidx = artist_enc.transform([artist_name])[0]
            artist_to_tracks[aidx].append(tid)

    recalls = []
    precisions = []
    for uid, test_tids in test_user_tracks.items():
        if uid not in user_id_to_idx:
            continue
        u = user_id_to_idx[uid]
        scores = user_emb[u].dot(item_emb.T)
        top_artists = np.argpartition(scores, -n_artists)[-n_artists:]
        top_artists = top_artists[np.argsort(scores[top_artists])[::-1]]

        seen = seen_tracks.get(uid, set())
        candidates = []
        for aidx in top_artists:
            track_list = artist_to_tracks.get(aidx, [])
            track_list_sorted = sorted(track_list, key=lambda t: track_pop[track_id_to_idx[t]], reverse=True)
            for t in track_list_sorted[:tracks_per_artist]:
                if t not in seen:
                    candidates.append((t, track_pop[track_id_to_idx[t]]))
        # deduplicate
        uniq = {}
        for t, pop in candidates:
            uniq[t] = pop
        candidates = sorted(uniq.items(), key=lambda x: x[1], reverse=True)
        top_tracks = [t for t,_ in candidates[:top_k]]

        hit = len(set(top_tracks) & test_tids)
        recall = hit / len(test_tids) if test_tids else 0.0
        precision = hit / top_k
        recalls.append(recall)
        precisions.append(precision)

    avg_recall = np.mean(recalls) if recalls else 0.0
    avg_precision = np.mean(precisions) if precisions else 0.0
    return avg_recall, avg_precision

# -------------------------------------------------------------------
# 9. Main
# -------------------------------------------------------------------
def main():
    # Load data (adjust to your files)
    # df_tracks = pd.read_csv('tracks.csv')
    # df_users = pd.read_csv('users.csv')

    print("Building user-artist matrix...")
    ua_mat, user_id_to_idx, artist_enc, track_to_artist = build_user_artist_matrix(df_tracks, df_users)
    print(f"User-Artist matrix: {ua_mat.shape[0]} users x {ua_mat.shape[1]} artists, {ua_mat.nnz} interactions")

    print("Building track test set...")
    test_user_tracks, seen_tracks, track_pop, track_id_to_idx = build_track_split(df_tracks, TEST_PERCENT)

    print("Training user-artist model (PyTorch BPR)...")
    user_emb, artist_emb = train_user_artist_model(ua_mat, EPOCHS)

    print("Evaluating track recall & precision...")
    rec, prec = evaluate_track_recall_and_precision(
        user_emb, artist_emb, user_id_to_idx, artist_enc,
        track_to_artist, test_user_tracks, seen_tracks,
        track_pop, track_id_to_idx
    )
    print(f"Track Recall@10: {rec:.4f}   Precision@10: {prec:.4f}")

if __name__ == "__main__":
    main()
```
